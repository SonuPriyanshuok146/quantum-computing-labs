```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
from math import pi
```

```python
sim = AerSimulator(seed_simulator=2024)

def run_and_plot(qc, shots=1024,title=None):
    counts = sim.run(transpile(qc, sim), shots=shots).result().get_counts()
    plot_histogram(counts, title=title)
    return counts
```

```python
qc_ghz = QuantumCircuit(3)
qc_ghz.h(0); qc_ghz.cx(0, 1); qc_ghz.cx(0, 2)
qc_ghz.measure_all()
print("GHZ:", run_and_plot(qc_ghz, 4000, "GHZ"))
```

```python
qc_z = QuantumCircuit(1); qc_z.h(0); qc_z.measure_all()
print("Z basis:", run_and_plot(qc_z))                    # ~50/50
qc_x = QuantumCircuit(1); qc_x.h(0); qc_x.h(0); qc_x.measure_all()
print("X basis:", run_and_plot(qc_x))   
```

```python
qc_sw = QuantumCircuit(2)
qc_sw.ry(pi/3, 0); qc_sw.rx(pi/5, 0)
qc_sw.cx(0, 1); qc_sw.cx(1, 0); qc_sw.cx(0, 1)
qc_sw.measure_all()
print("SWAP:", run_and_plot(qc_sw, 2000)) 
```

```python
def bell(variant):
    qc = QuantumCircuit(2)
    if variant in ("phi-", "psi-"): qc.x(0)              # start control in |->
    qc.h(0)
    if variant in ("psi+", "psi-"): qc.x(1)              # flip target to |1>
    qc.cx(0, 1)
    qc.measure_all()
    return qc
for v in ["phi+", "phi-", "psi+", "psi-"]:
    print(v, run_and_plot(bell(v), title=v))
```
