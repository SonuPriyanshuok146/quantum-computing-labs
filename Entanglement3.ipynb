```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import Aer
from qiskit.visualization import plot_histogram, plot_state_qsphere, plot_bloch_multivector
```

```python
qc_bell = QuantumCircuit(2)
qc_bell.h(0)          # superposition on qubit 0
qc_bell.cx(0, 1) 
```

```python
sv = Aer.get_backend('statevector_simulator')
psi = sv.run(transpile(qc_bell, sv)).result().get_statevector(qc_bell)
print("Bell state:", psi)                   # [0.707, 0, 0, 0.707]
plot_state_qsphere(psi)
```

```python
un = Aer.get_backend('unitary_simulator')
print(un.run(transpile(qc_bell, un)).result().get_unitary(qc_bell, decimals=3))
```

```python
# Measure both qubits
qc_m = qc_bell.copy()
qc_m.measure_all()
qasm = Aer.get_backend('qasm_simulator')
counts = qasm.run(transpile(qc_m, qasm), shots=1000).result().get_counts()
print("Bell counts:", counts)               # only '00' and '11', ~500 each
plot_histogram(counts)
```

```python
# Separable comparison (no CNOT)
qc_sep = QuantumCircuit(2)
qc_sep.h(0)
print("Separable state:", sv.run(transpile(qc_sep, sv)).result().get_statevector(qc_sep))
```
