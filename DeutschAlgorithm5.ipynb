```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import Aer
from qiskit.visualization import plot_histogram
```

```python
def deutsch(oracle):
    qc = QuantumCircuit(2, 1)
    qc.x(1)                       # ancilla to |1>
    qc.h(0); qc.h(1)              # |+>|->
    qc.barrier()
    if oracle == "const0":
        pass                      # f(x)=0
    elif oracle == "const1":
        qc.x(1)                   # f(x)=1
    elif oracle == "identity":
        qc.cx(0, 1)               # f(x)=x   (balanced)
    elif oracle == "negation":
        qc.cx(0, 1); qc.x(1)      # f(x)=NOT x (balanced)
    qc.barrier()
    qc.h(0)
    qc.measure(0, 0)
    return qc
```

```python
backend = Aer.get_backend('qasm_simulator')
for o in ["const0", "const1", "identity", "negation"]:
    qc = deutsch(o)
    counts = backend.run(transpile(qc, backend), shots=1000).result().get_counts()
    
    verdict = "constant" if '0' in counts and len(counts) == 1 else "balanced"
    print(o, counts, "->", verdict)
```
