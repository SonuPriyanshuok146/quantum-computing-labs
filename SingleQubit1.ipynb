```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import Aer
from qiskit.visualization import plot_histogram, plot_bloch_multivector
```

```python
qc_x = QuantumCircuit(1)
qc_x.x(0)
sv = Aer.get_backend('statevector_simulator')
state = sv.run(transpile(qc_x, sv)).result().get_statevector(qc_x, decimals=3)

print("State Vector after X:", state)
plot_bloch_multivector(state)
```

```python
qc_m = QuantumCircuit(1, 1)
qc_m.x(0)
qc_m.measure(0, 0)
qasm = Aer.get_backend('qasm_simulator')
counts = qasm.run(transpile(qc_m, qasm), shots=1024).result().get_counts()
print("Counts after X: ",counts)
plot_histogram(counts)
```

```python
qc_h = QuantumCircuit(1, 1)
qc_h.h(0)
qc_h.measure(0, 0)
counts_h = qasm.run(transpile(qc_h, qasm), shots=1000).result().get_counts()
print("Counts after H:", counts_h)  
```
