```python
from qiskit import QuantumCircuit
from qiskit_aer import Aer
from qiskit.visualization import plot_histogram
```

```python
def create_bell_pair(qc, a, b):
    qc.h(a)
    qc.cx(a, b)

def encode_message(qc, qubit, msg):
    if msg == "00":
        pass                       # do nothing
    elif msg == "10":
        qc.x(qubit)
    elif msg == "01":
        qc.z(qubit)
    elif msg == "11":
        qc.z(qubit); qc.x(qubit)
    else:
        print("Invalid message: sending '00'")
        
```

```python
def decode_message(qc, a, b):
    qc.cx(a, b)
    qc.h(a)
def run_superdense_coding(message):
    qc = QuantumCircuit(2, 2)
    create_bell_pair(qc, 0, 1)     # Charlie
    qc.barrier()
    encode_message(qc, 0, message) # Alice
    qc.barrier()
    decode_message(qc, 0, 1)       # Bob
    qc.measure([0, 1], [0, 1])
    backend = Aer.get_backend('qasm_simulator')
    counts = backend.run(qc, shots=1024).result().get_counts(qc)
    print(f"Sent {message} -> received {counts}")
    return counts
```

```python
for msg in ["00", "01", "10", "11"]:
    run_superdense_coding(msg)
```
